```python
from langchain_groq import ChatGroq
import json
def extract_cv():
  base_instruction = (
      "Extract the following information from the CV: name, email, phone, job title, "
      "profile/summary, education, experience, projects, and languages. "
      "Return the result ONLY in valid JSON nothing more with this exact structure:\n\n"
      "{\n"
      '  "name": "...",\n'
      '  "mail": "...",\n'
      '  "phone": "...",\n'
      '  "job": "...",\n'
      '  "profile": "...",\n'
      '  "education": [\n'
      '    { "university": "...", "study": "..." }\n'
      "  ],\n"
      '  "projects": [\n'
      '    { "project": "...", "description": "..." }\n'
      "  ],\n"
      '  "experience": [\n'
      '    { "company": "...", "job_title": "...", "description": "..." }\n'
      "  ],\n"
      '  "languages": [\n'
      '    { "language": "...", "level": "..." }\n'
      "  ]\n"
      "}\n\n"
      "If a field is missing in the CV, return an empty string or empty array for that field."
      f"this is the cv :{cv}"
  )
  llm = ChatGroq(model="llama-3.1-8b-instant")
  response = llm.invoke(base_instruction)
  doc=json.loads(response.content)
  return doc

doc=extract_cv()
print(doc)
```

```python
import ast
state={'name':"",'mail':"",'phone':"",'job':"",'profile':"",'education':[],'projects':[],'experience':[],'languages':[]}
def profiale(state,doc):

  state["name"]=doc["name"]
  state["mail"]=doc["mail"]
  state["phone"]=doc["phone"]
  state["job"]=doc["job"]

  base_instruction = ("You are an AI that fixes and improves CV.\n"
                      "This is the user's summary/profile. Improve it and make it better make it the lenght of the profiale cv summary/profil.\n\n"
                      f"profile/summary: {doc['profile']}\n\n"
                      "Return only the improved summary don't add any comment.")

  llm = ChatGroq(model="llama-3.1-8b-instant")
  response = llm.invoke(base_instruction)
  state["profile"]=response.content

  return state

state=profiale(state,doc)
print(state)

```

```python
def experience(state,doc):
  base_instruction = ("You are an AI that fixes and improves CV.\n"
                      "This is the user's experience. Improve it and make it better make it .\n\n"
                      f"experience: {doc['experience']}\n\n"
                      "Return only the improved experience as the same structure as { 'company': '...', 'job_title': '...', 'description': '...' }  nothing more  don't add any comment.")

  llm = ChatGroq(model="llama-3.1-8b-instant")
  while True:
      try:
        response = llm.invoke(base_instruction)
        #print(response.content)

        #print(response.content.split('\n'))
        projects = response.content.split('\n')
        # Convert each string (Python-style dict) into a Python dict
        projects_dicts = [ast.literal_eval(p) for p in projects]

        # Convert Python list of dicts into JSON string
        projects_json = json.dumps(projects_dicts, indent=2)
        break
      except Exception as e:
        print(f"Error occurred: {e}. Please try again.")

  projects_json1=[]
  for l in ast.literal_eval(projects_json):
    if isinstance(l, list):
      projects_json1.append(l[0])
    else:
      projects_json1.append(l)

  if projects_json1!=[]:
    state["experience"]=projects_json1
  else:
    state["experience"]=ast.literal_eval(projects_json)
  return state

state=experience(state,doc)

print(state)
```

```python
def languages(state,doc):
  base_instruction = ("You are an AI that fixes and improves CV.\n"
                      "This is the user's languages. Improve it and make it better make it .\n\n"
                      f"languages: {doc['languages']}\n\n"
                      "Return only the improved languages as the same structure as { 'language': '...', 'level': '...' }  nothing more  don't add any comment.")

  llm = ChatGroq(model="llama-3.1-8b-instant")
  while True:
      try:
        response = llm.invoke(base_instruction)
        projects = response.content.split('\n')
        # Convert each string (Python-style dict) into a Python dict
        projects_dicts = [ast.literal_eval(p) for p in projects]

        # Convert Python list of dicts into JSON string
        projects_json = json.dumps(projects_dicts, indent=2)
        break
      except Exception as e:
        print(f"Error occurred: {e}. Please try again.")

  #print(ast.literal_eval(projects_json))
  projects_json1=[]

  for l in ast.literal_eval(projects_json):
    if isinstance(l, list):
      projects_json1.append(l[0])
    else:
      projects_json1.append(l)

  if projects_json1!=[]:
    state["languages"]=projects_json1
  else:
    state["languages"]=ast.literal_eval(projects_json)
  return state

state=languages(state,doc)
print(state)
```

```python
def cv_to_latex(state):
  # Example data
  data=state
  # Start LaTeX document
  latex = r"""\documentclass[10pt, letterpaper]{article}

  % Packages:
  \usepackage[
      ignoreheadfoot,
      top=2 cm,
      bottom=2 cm,
      left=2 cm,
      right=2 cm,
      footskip=1.0 cm
  ]{geometry}
  \usepackage[explicit]{titlesec}
  \usepackage{tabularx}
  \usepackage{array}
  \usepackage[dvipsnames]{xcolor}
  \definecolor{primaryColor}{RGB}{0, 79, 144}
  \usepackage{enumitem}
  \usepackage{fontawesome5}
  \usepackage{amsmath}
  \usepackage[
      pdftitle={{ name }},
      pdfauthor={{ name }},
      pdfcreator={LaTeX with RenderCV},
      colorlinks=true,
      urlcolor=primaryColor
  ]{hyperref}
  \usepackage[pscoord]{eso-pic}
  \usepackage{calc}
  \usepackage{bookmark}
  \usepackage{lastpage}
  \usepackage{changepage}
  \usepackage{paracol}
  \usepackage{ifthen}
  \usepackage{needspace}
  \usepackage{iftex}

  \ifPDFTeX
      \input{glyphtounicode}
      \pdfgentounicode=1
      \usepackage[T1]{fontenc}
      \usepackage[utf8]{inputenc}
      \usepackage{lmodern}
  \fi

  \usepackage[default, type1]{sourcesanspro}

  \AtBeginEnvironment{adjustwidth}{\partopsep0pt}
  \pagestyle{empty}
  \setcounter{secnumdepth}{0}
  \setlength{\parindent}{0pt}
  \setlength{\topskip}{0pt}
  \setlength{\columnsep}{0.15cm}
  \makeatletter
  \let\ps@customFooterStyle\ps@plain
  \patchcmd{\ps@customFooterStyle}{\thepage}{
  }{}{}
  \makeatother
  \pagestyle{customFooterStyle}

  \titleformat{\section}{
      \needspace{4\baselineskip}
      \Large\color{primaryColor}
  }{
  }{
  }{
      \textbf{#1}\hspace{0.15cm}\titlerule[0.8pt]\hspace{-0.1cm}
  }[]

  \titlespacing{\section}{
      -1pt
  }{
      0.3 cm
  }{
      0.2 cm
  }

  \newenvironment{highlights}{
      \begin{itemize}[
          topsep=0.10 cm,
          parsep=0.10 cm,
          partopsep=0pt,
          itemsep=0pt,
          leftmargin=0.4 cm + 10pt
      ]
  }{
      \end{itemize}
  }

  \newenvironment{highlightsforbulletentries}{
      \begin{itemize}[
          topsep=0.10 cm,
          parsep=0.10 cm,
          partopsep=0pt,
          itemsep=0pt,
          leftmargin=10pt
      ]
  }{
      \end{itemize}
  }

  \newenvironment{onecolentry}{
      \begin{adjustwidth}{
          0.2 cm + 0.00001 cm
      }{
          0.2 cm + 0.00001 cm
      }
  }{
      \end{adjustwidth}
  }

  \newenvironment{twocolentry}[2][]{
      \onecolentry
      \def\secondColumn{#2}
      \setcolumnwidth{\fill, 4.5 cm}
      \begin{paracol}{2}
  }{
      \switchcolumn \raggedleft \secondColumn
      \end{paracol}
      \endonecolentry
  }

  \newenvironment{threecolentry}[3][]{
      \onecolentry
      \def\thirdColumn{#3}
      \setcolumnwidth{1 cm, \fill, 4.5 cm}
      \begin{paracol}{3}
      {\raggedright #2} \switchcolumn
  }{
      \switchcolumn \raggedleft \thirdColumn
      \end{paracol}
      \endonecolentry
  }

  \newenvironment{header}{
      \setlength{\topsep}{0pt}\par\kern\topsep\centering\color{primaryColor}\linespread{1.5}
  }{
      \par\kern\topsep
  }

  \newcommand{\placelastupdatedtext}{
    \AddToShipoutPictureFG*{
      \put(
          \LenToUnit{\paperwidth-2 cm-0.2 cm+0.05cm},
          \LenToUnit{\paperheight-1.0 cm}
      ){\vtop{{\null}\makebox[0pt][c]{
          \small\color{gray}\textit{}\hspace{\widthof{Last updated in September 2024}}
      }}}%
    }%
  }

  \let\hrefWithoutArrow\href
  \renewcommand{\href}[2]{\hrefWithoutArrow{#1}{\ifthenelse{\equal{#2}{}}{ }{#2 }\raisebox{.15ex}{\footnotesize \faExternalLink*}}}

  \begin{document}
  \newcommand{\AND}{\unskip
      \cleaders\copy\ANDbox\hskip\wd\ANDbox
      \ignorespaces
  }
  \newsavebox\ANDbox
  \sbox\ANDbox{}

  \placelastupdatedtext
  \begin{header}
      \fontsize{30 pt}{30 pt}
      \textbf{ """ + data["name"] + r""" }

      \vspace{0.3 cm}

      \normalsize
      \mbox{\hrefWithoutArrow{mailto:""" + data["mail"] + r"""}{{\footnotesize\faEnvelope[regular]}\hspace*{0.13cm}""" + data["mail"] + r"""}}%
      \kern 0.25 cm%
      \AND%
      \kern 0.25 cm%
      \mbox{\hrefWithoutArrow{tel:""" + data["phone"] + r"""}{{\footnotesize\faPhone*}\hspace*{0.13cm}""" + data["phone"] + r"""}}%
  \end{header}

  \vspace{0.3 cm - 0.3 cm}

  \section{Profile}
  \begin{onecolentry}
  """ + data["profile"] + r"""
  \end{onecolentry}

  \vspace{0.2 cm}

  \section{Education}
  """

  # Education
  for edu in data["education"]:
      latex += r"""
  \begin{threecolentry}{}{
  }
      \textbf{ """ + edu["university"] + r""" }
      \begin{highlights}
          \item """ + edu["study"] + r"""
      \end{highlights}
  \end{threecolentry}
  """

  # Experience
  latex += r"\section{Experience}"
  for exp in data["experience"]:
      latex += r"""
  \begin{twocolentry}{
  }
      \textbf{ """ + exp["company"] + r""" }, """ + exp["job_title"] + r"""
      \begin{highlights}
          \item """ + exp["description"] + r"""
      \end{highlights}
  \end{twocolentry}

  \vspace{0.2 cm}
  """

  # Projects
  latex += r"\section{Projects}"
  for pro in data["projects"]:
      latex += r"""
  \begin{twocolentry}{}
      \textbf{ """ + pro["project"] + r""" }
      \begin{highlights}
          \item """ + pro["description"] + r"""
      \end{highlights}
  \end{twocolentry}

  \vspace{0.2 cm}
  """

  # Languages
  latex += r"\section{Languages}"
  for lang in data["languages"]:
      latex += r"""
  \begin{onecolentry}
      \textbf{ """ + lang["language"] + r""" :} """ + lang["level"] + r"""
  \end{onecolentry}

  \vspace{0.2 cm}
  """

  latex += r"\end{document}"

  # Save LaTeX file
  with open("cv.tex", "w", encoding="utf-8") as f:
      f.write(latex)

  print("✅ cv.tex created successfully!")

cv_to_latex(state)
```

```python
import subprocess

def run_latex():
    result = subprocess.run(
        ["pdflatex", "-interaction=nonstopmode", "cv.tex"],
        stdout=subprocess.PIPE,
        stderr=subprocess.PIPE,
        text=True
    )
    #print(result.stdout)   # LaTeX output
    #print(result.stderr)   # Errors
    return result.returncode

run_latex()
```
